# Me and My Ten Rooms, in a notebook

A runnable companion to the ilm.red post **"Me and my ten rooms"**. Alma walked through ten rooms,
each one a philosophical position taken completely seriously. This notebook walks the same corridor
in code: each room becomes a tiny, honest simulation, small enough to read whole, and each one breaks
in exactly the way the room did.

No accounts, no API keys, standard library only. Run it top to bottom.


```python
# Setup: standard library only. One shared die, so every room is replayable.
import random, statistics, math
from collections import Counter
rng = random.Random(10)   # ten rooms, seed 10
print('the corridor is open')
```

## Room 1 · The doorman (Nihilism)

A value function that returns zero for everything. Nothing is worth more than anything else,
so every plan is exactly as good as every other plan, including the plan of standing still.
Watch an optimizer try to choose.


```python
plans = ['study all night', 'sleep', 'burn the textbook', 'stand perfectly still']
def nihilist_value(plan):
    return 0.0    # the doorman shrugs

best = max(plans, key=nihilist_value)
print('every plan scores', {p: nihilist_value(p) for p in plans})
print('the optimizer "chose":', repr(best), '-- max() just returns the first tie.')
print('That is not a decision. It is the order the list happened to be written in.')
```

## Room 2 · The tortoise in the rain (Stoicism)

The dichotomy of control: the rain is not up to you, your response is. Two agents live through
the same 1,000 random days. One spends effort fighting the weather; one spends the same effort
only on what it controls.


```python
def day(): return rng.random() < 0.4          # 40% of days it rains; nobody controls this
def fighter(rain):   return -2.0 if rain else 1.0   # rages at rain it cannot stop
def tortoise(rain):  return  0.5                     # tends its rock either way
days = [day() for _ in range(1000)]
print('fighter  mean mood:', round(statistics.mean(fighter(r) for r in days), 3))
print('tortoise mean mood:', round(statistics.mean(tortoise(r) for r in days), 3))
print('Same weather. The only variable either agent ever controlled was the response function.')
```

## Room 3 · The silk dressing gown (Hedonism)

The lemon drizzle is exquisite -- the first slice. Pleasure adapts: the same stimulus, repeated,
pays less each time. This curve has a name in economics and a couch in this room.


```python
def pleasure(slice_n, adaptation=0.6):
    return adaptation ** (slice_n - 1)      # each slice is worth 60% of the one before
total = 0.0
for n in range(1, 11):
    total += pleasure(n)
    print(f'slice {n:2d}: tastes like {pleasure(n):.3f}   running total {total:.3f}')
print('Ten slices never reach 2.5. The treadmill converges; the appetite does not.')
```

## Room 4 · The tea party (Solipsism)

The man is sure this is his dream. Can he check? Two worlds: one real, one simulated from inside
his own head. Every observation he can make passes through the same senses -- so the two streams
are byte-identical, and no experiment run from inside can split them.


```python
def real_world(n):      return [random.Random(42+i).random() for i in range(n)]
def dreamed_world(n):   return [random.Random(42+i).random() for i in range(n)]  # same generator: HIS head
a, b = real_world(8), dreamed_world(8)
print('real   :', [round(x,3) for x in a])
print('dreamed:', [round(x,3) for x in b])
print('distinguishable?', a != b)
print('The position is irrefutable from inside, which is different from being true, and')
print('exactly why it is unfalsifiable rather than useful.')
```

## Room 5 · The grinning cat (Relativism)

The cat agrees with everything. In information terms an evaluator that returns the same verdict
for every claim carries **zero bits**: you learn nothing from asking it. Entropy of its answers, measured.


```python
claims = ['the exam is tomorrow', 'the exam was yesterday', '2+2=4', '2+2=5']
def cat(claim): return 'yes, quite true'
answers = Counter(cat(c) for c in claims)
probs = [n/len(claims) for n in answers.values()]
H = abs(-sum(p*math.log2(p) for p in probs))
print('distinct answers:', dict(answers))
print(f'entropy of the cat as an information source: {H:.1f} bits')
print('An oracle that cannot disagree cannot inform. If every view is equally right,')
print('asking is decoration.')
```

## Room 6 · The measuring mole (Empiricism)

The mole trusts only measurement. Measurement is genuinely powerful -- and it has a blind spot the
size of tomorrow: every observation is about the past, and the leap to the next case is not itself
a measurement. The mole meets a feeding schedule.


```python
# A chicken is fed on 999 consecutive mornings. Estimate P(fed) empirically.
observations = [1.0] * 999
p_hat = sum(observations) / len(observations)
print(f'empirical P(breakfast) after 999 mornings: {p_hat:.3f}')
day_1000 = 'the cook wrings its neck'   # Russell's chicken, exactly on schedule
print('day 1000:', day_1000)
print('The data was perfect. The inference rule -- tomorrow resembles yesterday -- was the')
print('unmeasured assumption doing all the work. That rule is not itself in the data.')
```

## Room 7 · The spider's web (Rationalism)

The web hangs from a few anchor threads and everything else is derived, flawlessly. Deduction
really is watertight -- **given the anchors**. Move one axiom a little and the whole web moves
with it, still internally perfect.


```python
def web(anchor_interior_angles):
    # 'derive' the angle sum of a triangle from the chosen geometry axiom
    return {'flat plane (Euclid)': 180, 'sphere (riemannian)': 'more than 180',
            'saddle (hyperbolic)': 'less than 180'}[anchor_interior_angles]
for axiom in ['flat plane (Euclid)', 'sphere (riemannian)', 'saddle (hyperbolic)']:
    print(f'axiom: {axiom:24s} -> triangle angles sum to {web(axiom)}')
print('Three perfect webs, three different worlds. Deduction transmits certainty;')
print('it cannot manufacture it. Someone still has to check which anchors hold in THIS room.')
```

## Room 8 · The clockwork courtier (Determinism)

Same mechanism, same starting state, same history -- run it twice and watch. Then move the
starting state by one part in a million and watch a different court unfold. Determined is not
the same as predictable.


```python
def court(seed, steps=15):
    r = random.Random(seed)
    x = 0.4
    for _ in range(steps):
        x = 3.99 * x * (1 - x)      # a deterministic map, famously touchy
    return round(x, 6)
print('run A, state 0.4        :', court(1))
print('run B, same everything  :', court(1))
def court_from(x0, steps=15):
    x = x0
    for _ in range(steps): x = 3.99 * x * (1 - x)
    return round(x, 6)
print('start 0.400000          :', court_from(0.400000))
print('start 0.400001          :', court_from(0.400001))
print('The courtier is a mechanism AND his future is practically unknowable. The room')
print('quietly conflated those, and the price was pinned on nobody.')
```

## Room 9 · The Queen's spreadsheet (Utilitarianism)

The greatest good for the greatest number, as an actual spreadsheet. The optimizer below is not
evil; it is obedient. Watch what the argmax does to the smallest row.


```python
village = {'Ada': 6, 'Bo': 5, 'Cy': 5, 'Din': 5, 'the doorman': 2}   # current welfare
def total(v): return sum(v.values())
def policy_share(v):      return {k: w + 1 for k, w in v.items()}
def policy_harvest(v):    # convert the worst-off row into +2 for everyone else
    worst = min(v, key=v.get)
    return {k: (0 if k == worst else w + 2) for k, w in v.items()}
for name, pol in [('share with all', policy_share), ('harvest the worst-off', policy_harvest)]:
    out = pol(village)
    print(f'{name:22s} total {total(out):3d}   {out}')
print('The spreadsheet prefers the harvest. Nothing in the objective says it may not;')
print('the constraint you wanted was never a cell in it.')
```

## Room 10 · The chair, the note, and the choice (Existentialism)

After nine rooms of inherited answer keys, the last room contains no objective function at all.
The agent must supply one -- and THAT choice, which no room can make for it, decides everything
downstream. Existence precedes the essence; in code: the policy exists before its utility is chosen.


```python
corridors = {
  'maximize total':    lambda v: sum(v.values()),
  'maximize the floor':lambda v: min(v.values()),   # Rawls sits quietly in this corner
  'maximize my row':   lambda v: v['Ada'],
}
options = [ {'Ada': 6,'Bo': 5,'Cy': 5,'Din': 5,'the doorman': 2},
            {'Ada': 4,'Bo': 4,'Cy': 4,'Din': 4,'the doorman': 4},
            {'Ada': 9,'Bo': 1,'Cy': 1,'Din': 1,'the doorman': 1} ]
for name, U in corridors.items():
    pick = max(options, key=U)
    print(f'{name:20s} -> chooses {pick}')
print()
print('Three sincere agents, three worlds. The utility function was not found in any room:')
print('it was chosen. Owning that choice, instead of outsourcing it to a doorman, a cat or a')
print('spreadsheet, is the exam the corridor was actually setting.')
```
